## Project Overview

Urban flooding is a growing threat to Southeast Asian cities, risking lives and economic stability. **Transportation resilience** plays a crucial role in this context, as it refers to a city’s ability to keep its mobility systems functioning during uncertain times. When road networks fail to coordinate, disaster management becomes ineffective, amplifying socio-economic losses.

This project focuses on **Hat Yai**, one of Thailand’s key economic centers, to analyze a recent major flood. We evaluate how the flood affected transportation and the extent of human exposure. Based on these findings, we aim to build a framework to classify flood impacts and provide insights to improve future urban planning and emergency response.

## Data Sources
This analysis relies on open-source data from:

* **Digital Elevation Model (DEM):** [NASA Earthdata](https://dwtkns.com/srtm30m/)
    * *Used for topographical analysis and "Bathtub Model" flood simulation.*
* **Road Network:** [OpenStreetMap (OSM)](https://www.openstreetmap.org/) via [OSMnx](https://osmnx.readthedocs.io/)
    * *Used for graph network construction and connectivity analysis.*
* **Flood Extent:** [Sentinel Asia](https://sentinel-asia.org/EO/2025/article20251119TH.html)
    * *Historical flood footprint data (Nov 23, 2025).*
* **Population:** [WorldPop](https://hub.worldpop.org/geodata/summary?id=56706)
    * *High-resolution (100m) gridded population density data.*
