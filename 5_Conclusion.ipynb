# Conclusion & Future Outlook

## Synthesis of Findings

The multi-dimensional analysis confirms that Hat Yai faces a systemic resilience deficit, revealing a perilous convergence of risks where a transport network relies heavily on a vulnerable geographic core situated within an urban topography lacking hydraulic buffers. Topologically, the network analysis demonstrates that flood-induced disruptions are non-linear; the loss of critical central nodes triggers a massive surge in systemic stress (+86.4%), forcing traffic onto an inadequate peripheral grid and creating functional "islands". Topographically, the scenario analysis highlights that the threshold for disaster is dangerously low, as a minor rise of just one meter is sufficient to paralyze nearly a quarter of the road network and impact 30% of the population, indicating that historical urban expansion has occurred within high-risk hydrological depressions.

## Methodological Limitations

While this study provides a robust framework for rapid vulnerability assessment, several methodological constraints must be acknowledged. First, the reliance on a static model assumes a planar water surface based solely on elevation, omitting hydrodynamic factors such as flow velocity and existing flood defenses like pumps. Second, the graph analysis treats road segments using uniform weights based primarily on distance, without accounting for road capacity or pre-existing congestion levels. Finally, the population exposure metrics are derived from static residential data, meaning they do not capture diurnal movements such as commuters and students, which could significantly alter the actual population at risk depending on the time of day.


## Future Scope

Ultimately, this study serves as a strategic imperative for urban planners and policymakers to shift from reactive disaster response to proactive resilience planning. The findings underscore the urgent need to move beyond historical expansion patterns that ignore hydrological risks and instead integrate land-use planning directly with flood management strategies. Future governance must prioritize the development of a "redundant" transportation grid that does not rely solely on the vulnerable city center but actively strengthens peripheral connectivity to prevent systemic paralysis. By adopting a data-driven approach to infrastructure investment and strictly regulating high-density development in high-risk depressions, authorities can ensure that Hat Yai’s role as a regional economic engine remains secure, even in the face of escalating climate uncertainty.
