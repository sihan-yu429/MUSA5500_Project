# Flood Exposure Analysis 

## Case Event: The November 23, 2025 Flood

### Hydrological Magnitude and Impact
Precipitated by an intense monsoon trough and a severe low-pressure system commencing on November 21, 2025, this event paralleled the spatial scale of the historic 2000 flood but exceeded it in total hydrological volume. The widespread inundation directly affected approximately 104,917 households, with preliminary economic damages estimated to surpass 500 million. Tragically, as of the time of writing, the disaster has claimed more than 100 lives in the hard-hit Songkhla province.

### Systemic Failure Analysis
The disaster exposed critical limitations in Hat Yai’s flood mitigation infrastructure. The sheer volume of runoff overwhelmed the design capacity of the U-Tapao and Phuminatdanri diversion canals, which were unable to manage simultaneous converging flows from three upstream tributaries. This vulnerability was exacerbated by two key factors: rapid urbanization, which has drastically reduced natural retention zones over the last decade, and intensifying rainfall trends (observed 2022–2025). Current data suggests that existing defense systems are functionally insufficient to cope with these escalating climatic extremes.

> **Note on Temporal Scope:** While the flood event remains active, this analysis utilizes the satellite imagery acquired on **November 23, 2025**, as the primary data snapshot. This specific timeframe provides the most current verifiable extent for assessing immediate infrastructure and population exposure.

## Inundated Network 
To understand the structural causes behind such catastrophic disruption, here first visualized the inundated network in the city.


```python
import geopandas as gpd
import osmnx as ox
import matplotlib.pyplot as plt


flood_path = "data/EOS-RS_20251109_FPM_S1_Philippines_Bicol_Region_Typhoon_Fung-Wong_v0.9_shp/EOS-RS_20251123_FPM_S1_Thailand_Hat_Yai_Floods_v0.9_shp.shp"
try:
    flood_gdf = gpd.read_file(flood_path)
except Exception as e:
    print(f"cannot find, please check: {e}")


if flood_gdf.crs is None:
    flood_gdf.set_crs(epsg=4326, inplace=True)
flood_gdf = flood_gdf.to_crs(epsg=32647)



if 'edges' not in globals():
 
    place_name = "Hat Yai, Songkhla, Thailand"
    G = ox.graph_from_place(place_name, network_type='drive')
    nodes, edges = ox.graph_to_gdfs(G)
    edges = edges.to_crs(epsg=32647)


if 'hat_yai_city' not in globals():
    hat_yai_city = ox.geocode_to_gdf("Hat Yai, Songkhla, Thailand").to_crs(epsg=32647)


flood_clean = gpd.clip(flood_gdf, hat_yai_city)

flood_clean = flood_clean[flood_clean.geometry.area > 100]


fig, ax = plt.subplots(figsize=(12, 12)) 


hat_yai_city.plot(ax=ax, color='none', edgecolor='black', linewidth=1.5, label='City Boundary', zorder=3)


edges.plot(ax=ax, color='#999999', linewidth=0.5, alpha=0.5, label='Road Network', zorder=1)


flood_clean.plot(ax=ax, color='#7CD0E6', alpha=0.7, label='Flood Extent (Nov 23)', zorder=2)

plt.title(f"Flood Exposure: Hat Yai Road Network\n(Nov 23, 2025)", fontsize=15)
 
from matplotlib.lines import Line2D
from matplotlib.patches import Patch
custom_lines = [
    Line2D([0], [0], color='#3B3331', lw=2, label='Boundary'),
    Line2D([0], [0], color='#999999', lw=1, label='Roads'),
    Patch(facecolor='#7CD0E6', edgecolor='none', label='Flood Water')
]
ax.legend(handles=custom_lines, loc='upper right')

# 去掉坐标轴
ax.axis('off')

plt.show()
```

The spatial configuration of the flood extent (Blue) versus the transportation network (Gray) reveals a critical vulnerability: inundation is not randomly distributed but is heavily concentrated in the central and northern districts—precisely where road density is highest. By isolating the submerged segments (marked in Orange), we calculate that approximately 22.8% of the total road network (1,058 km out of 4,631 km) was directly submerged. Crucially, this statistic understates the functional severity; while 77% of the roads technically remain "dry," the connectivity of the entire system is severely compromised because the flooded portion represents the city's central logistics hub, effectively paralyzing movement across the wider metropolitan area.

```python
import geopandas as gpd
import matplotlib.pyplot as plt


COLORS = {
    'background': '#FFFFFF',
    'boundary': '#212121',
    'road_safe': '#999999',     
    'road_flooded': '#7CD0E6', 
}


flooded_roads = gpd.overlay(edges, flood_clean, how='intersection')

total_len_km = edges.length.sum() / 1000
flooded_len_km = flooded_roads.length.sum() / 1000
percent_impact = (flooded_len_km / total_len_km) * 100


fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 8), gridspec_kw={'width_ratios': [2, 1]})

#Left
edges.plot(ax=ax1, color=COLORS['road_safe'], linewidth=0.5, zorder=1, label='Functional Roads')

flooded_roads.plot(ax=ax1, color=COLORS['road_flooded'], linewidth=1.5, zorder=2, label='Inundated Roads')

if 'hat_yai_city' in globals():
    hat_yai_city.plot(ax=ax1, facecolor='none', edgecolor=COLORS['boundary'], linewidth=2, zorder=3)

ax1.set_title(f"Impact Map: {percent_impact:.1f}% of Network Inundated", fontsize=14, fontweight='bold')
ax1.legend(loc='upper right')
ax1.axis('off')

# Right
bars = ax2.bar(['Safe Network', 'Flooded'], 
               [total_len_km - flooded_len_km, flooded_len_km], 
               color=[COLORS['road_safe'], COLORS['road_flooded']])


for bar in bars:
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height,
             f'{height:.1f} km',
             ha='center', va='bottom', fontsize=12)

ax2.set_title("Quantitative Assessment (km)", fontsize=14)
ax2.spines['top'].set_visible(False)
ax2.spines['right'].set_visible(False)
ax2.set_ylabel("Length (km)")

plt.tight_layout()
plt.show()
```

## Population at Risk 

Beyond infrastructure damage, the analysis reveals an immediate and severe threat to human life. By explicitly excluding the uninhabited mudflats and vacant lands located in the upper-right periphery, the spatial overlay demonstrates a critical correlation between floodwaters (Light Blue) and high-density population clusters (Dark Orange), indicating that the disaster disproportionately impacted the city's residential and commercial core rather than uninhabited area.

Quantitatively, the exposure is significant. Out of a total estimated population of 597,303, approximately 215,477 residents, which is about 36.1% of the city resides directly within the inundated zones. This high exposure rate underscores a systemic urban planning failure, as historical expansion into low-lying, flood-prone basins has placed over a third of the population in the direct path of hydrological hazards.

```python
import rasterio
from rasterio.mask import mask
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import LogNorm
import geopandas as gpd
import osmnx as ox
from matplotlib.lines import Line2D
from matplotlib.patches import Patch


tif_path = "data/tha_pop_2025_CN_100m_R2024B_v1.tif" 
flood_path = "data/EOS-RS_20251109_FPM_S1_Philippines_Bicol_Region_Typhoon_Fung-Wong_v0.9_shp/EOS-RS_20251123_FPM_S1_Thailand_Hat_Yai_Floods_v0.9_shp.shp"


hat_yai_city = ox.geocode_to_gdf("Hat Yai, Songkhla, Thailand")
city_wgs84 = hat_yai_city.to_crs(epsg=4326)

try:
    flood_gdf = gpd.read_file(flood_path)
    if flood_gdf.crs is None:
        flood_gdf.set_crs(epsg=4326, inplace=True)
    flood_wgs84 = flood_gdf.to_crs(epsg=4326)
    flood_clean_wgs84 = gpd.clip(flood_wgs84, city_wgs84)
except Exception as e:
    print(f"❌ failed: {e}")
    flood_clean_wgs84 = gpd.GeoDataFrame(geometry=[], crs="EPSG:4326")


try:
    with rasterio.open(tif_path) as src:
        out_image, out_transform = mask(src, city_wgs84.geometry, crop=True)
        height, width = out_image[0].shape
        west = out_transform[2]
        north = out_transform[5]
        east = west + (width * out_transform[0])
        south = north + (height * out_transform[4])
        extent = [west, east, south, north]
        
        pop_data = out_image[0]
        pop_data = np.where(pop_data <= 0, np.nan, pop_data)


        fig, ax = plt.subplots(figsize=(12, 12), facecolor='white')

        # Layer 1: population
        im = ax.imshow(pop_data, 
                       cmap='Oranges',
                       norm=LogNorm(), 
                       extent=extent, 
                       alpha=0.9,
                       zorder=1)

        # Layer 2: flood
        if not flood_clean_wgs84.empty:
            flood_clean_wgs84.plot(ax=ax, 
                                   color='#7CD0E6', 
                                   alpha=0.8,      
                                   edgecolor='none',
                                   zorder=2)

        # Layer 3: boundary
        city_wgs84.plot(ax=ax, facecolor='none', edgecolor='#595B5C', linewidth=2, zorder=3)

        ax.set_title("Human Exposure: Population at Risk\n(Event: Nov 23, 2025)", 
                     fontsize=16, fontweight='bold', pad=20)
        ax.axis('off')

        cbar = plt.colorbar(im, ax=ax, fraction=0.035, pad=0.04)
        cbar.set_label('Population Density (Log Scale)')

        legend_elements = [
            Line2D([0], [0], color='black', lw=2, label='City Boundary'),
            Patch(facecolor='orange', edgecolor='none', label='Population Density'), 
            Patch(facecolor='#7CD0E6', edgecolor='none', alpha=0.7, label='Inundated Area') 
        ]
        ax.legend(handles=legend_elements, loc='upper right', frameon=False, fontsize=11)

        plt.tight_layout()
        plt.show()

except FileNotFoundError:
    print(f"cannot find files")
except Exception as e:
    print(f"error: {e}")
```

```python
import rasterio
from rasterio.mask import mask
import numpy as np
import matplotlib.pyplot as plt


city_wgs84 = hat_yai_city.to_crs(epsg=4326)
flood_clean_wgs84 = flood_clean.to_crs(epsg=4326)

tif_path = "data/tha_pop_2025_CN_100m_R2024B_v1.tif"

with rasterio.open(tif_path) as src:
   
    out_image_city, _ = mask(src, city_wgs84.geometry, crop=True)
    data_city = out_image_city[0]
    data_city = np.where(data_city < 0, 0, data_city)
    total_pop = np.nansum(data_city)

    if not flood_clean_wgs84.empty:

        out_image_flood, _ = mask(src, flood_clean_wgs84.geometry, crop=True)
        data_flood = out_image_flood[0]
        data_flood = np.where(data_flood < 0, 0, data_flood)
        
        affected_pop = np.nansum(data_flood)
    else:
        affected_pop = 0

    affected_ratio = (affected_pop / total_pop) * 100

    fig, ax = plt.subplots(figsize=(6, 6), facecolor='white') 
    sizes = [total_pop - affected_pop, affected_pop]
    labels = ['Safe Population', 'Affected Population']
    colors = ['#E0E0E0', '#FF9800'] 
    explode = (0, 0.1)  

    ax.pie(sizes, explode=explode, labels=labels, colors=colors,
           autopct='%1.1f%%', shadow=False, startangle=140, 
           textprops={'fontsize': 12, 'fontweight': 'bold', 'color': '#333333'})
    
    ax.set_title("Population Impact Ratio\n(Nov 23 2025)", fontsize=14, fontweight='bold')
    plt.show()
```
